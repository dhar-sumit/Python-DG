```python
# Deduce the percentage of developers who knew python in each country.

import pandas as pd

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')

# Drop NaN in LanguageWorkedWith and Country columns
df = df[['Country', 'LanguageWorkedWith']].dropna()

# Create a boolean column indicating if 'python' is in LanguageWorkedWith
df['KnowsPython'] = df['LanguageWorkedWith'].apply(lambda x: 'python' in str(x).lower().split(';'))

# Group by Country and calculate total developers and python developers
result = df.groupby('Country').agg(
    Total_devs=('KnowsPython', 'count'),
    Python_devs=('KnowsPython', 'sum')
).reset_index()

# Calculate percentage of python developers
result['Python_percentage'] = (result['Python_devs'] / result['Total_devs']) * 100

# Sort by Python_percentage in descending order
result_sorted = result.sort_values('Python_percentage', ascending=False).reset_index(drop=True)

# Save CSV
import os
os.makedirs('../output', exist_ok=True)
result_sorted.to_csv('../output/q2_developers_knew_python.csv', index=False)

# Display the top 10 countries with highest percentage of python developers
result_sorted.head(10)
```
