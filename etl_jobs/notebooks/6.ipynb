```python
# Generate the report for job and career satisfaction of developers based on their gender and continent.
    
import pandas as pd
import numpy as np
import country_converter as coco

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv', low_memory=False)

# Create a working dataframe
df = df[['Gender', 'Country', 'JobSat', 'CareerSat']]
df = df.dropna(subset=['Country', 'JobSat', 'CareerSat'])

# Normalize Gender values
def normalize_gender(val):
    if pd.isna(val):
        return 'Others'
    s = str(val).strip()
    if s in ['Man', 'Woman']:
        return s
    else:
        return 'Others'
    
df['Gender'] = df['Gender'].apply(normalize_gender)

#  Country to Continent mapping
cc = coco.CountryConverter()
df = df[df['Country'] != 'Other Country (Not Listed Above)']
df['Continent'] = cc.convert(names=df['Country'], to='continent', not_found='Unknown')

# Mapping satisfaction to numeric
satisfaction_map = {
    'Very satisfied': 5,
    'Slightly satisfied': 4,
    'Neither satisfied nor dissatisfied': 3,
    'Slightly dissatisfied': 2,
    'Very dissatisfied': 1
}
df['JobSat'] = df['JobSat'].map(satisfaction_map)
df['CareerSat'] = df['CareerSat'].map(satisfaction_map)

# Group by Gender & Continent, calculate mean satisfaction
result = df.groupby(['Continent', 'Gender']).agg({
    'JobSat': 'mean',
    'CareerSat': 'mean',
}).round(2).reset_index()

# Save CSV
import os
os.makedirs('../output', exist_ok=True)
result.to_csv('../output/q6_job_carrer_satisfaction.csv')

# Display top 10 rows
result.head(10)
```
