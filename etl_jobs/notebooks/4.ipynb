```python
#  Based on this survey, what will be the most desired programming language for the year 2020?

import pandas as pd

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')
df = df[['LanguageDesireNextYear']].dropna()

# Split and explode the languages
df['LangList'] = df['LanguageDesireNextYear'].astype(str).str.split(';')
df = df.explode('LangList')
df['LangList'] = df['LangList'].str.strip()

# Count occurrences of each language
lang_counts = df['LangList'].value_counts().reset_index()
lang_counts.columns = ['language','count']

# Save to CSV
import os
os.makedirs('../output', exist_ok=True)
lang_counts.to_csv('../output/q4_desired_lang_2020.csv', index=False)

# Display the top 10 desired languages
print("Top 10 desired languages for 2020:")
print(lang_counts.head(10))

# Identify the most desired language
most_desired = lang_counts.iloc[0]['language'] if not lang_counts.empty else None
print("\nMost desired language for 2020 (survey):", most_desired)
```
