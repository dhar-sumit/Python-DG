```python
# Q1
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')

# Cleaning function
def clean_age(value):
    if pd.isna(value):
        return np.nan
    # Younger than 5 years
    if "Younger" in str(value):
        val = str(value).split()
        if len(val) > 1 and val[-2].isdigit():
            return int(val[-2])-1
        return np.nan
    # Older than 85
    if "Older" in str(value):
        val = str(value).split()
        if len(val) > 1 and val[-1].isdigit():
            return int(val[-1])+1
        return np.nan
    try:
        return int(value)
    except:
        return np.nan

# Apply cleaning
df['Age1stCode_clean'] = df['Age1stCode'].apply(clean_age)

# Average age
average_age = df['Age1stCode_clean'].mean()
print("Average age when developers wrote first line of code:", round(average_age, 2))
```

```python
# Q2
import pandas as pd

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')

# Drop NaN in LanguageWorkedWith and Country
lang_df = df[['Country', 'LanguageWorkedWith']].dropna()

# Create a flag for Python
lang_df['KnowsPython'] = lang_df['LanguageWorkedWith'].apply(lambda x: 'python' in str(x).lower().split(';'))

# Group by country
result = lang_df.groupby('Country').agg(
    total_devs=('KnowsPython', 'count'),
    python_devs=('KnowsPython', 'sum')
).reset_index()

# Calculate percentage
result['python_percentage'] = (result['python_devs'] / result['total_devs']) * 100

# Sort by highest percentage
result_sorted = result.sort_values('python_percentage', ascending=False).reset_index(drop=True)

print(result_sorted.head(10))  # top 10 countries
```

```python
# Q3
import pandas as pd
import country_converter as coco

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')

# Keep only salary and country, drop missing
salary_df = df[['Country', 'ConvertedComp']].dropna()

salary_df = salary_df[salary_df['Country'] != 'Other Country (Not Listed Above)']

# Map country → continent
cc = coco.CountryConverter()
salary_df['Continent'] = cc.convert(names=salary_df['Country'], to='continent')

# Group by continent and calculate average salary
avg_salary_by_continent = salary_df.groupby('Continent').agg(
    average_salary=('ConvertedComp', 'mean')
).reset_index()

# Sort by highest salary
avg_salary_by_continent_sorted = avg_salary_by_continent.sort_values('average_salary', ascending=False).reset_index(drop=True)
print(avg_salary_by_continent_sorted.head(10))  # top 10 continents
```

```python
# Q4
desire = df[['Respondent', 'LanguageDesireNextYear']].dropna(subset=['LanguageDesireNextYear']).copy()

# Split and explode
desire['LangList'] = desire['LanguageDesireNextYear'].astype(str).str.split(';')
desire = desire.explode('LangList')
desire['LangList'] = desire['LangList'].str.strip()

# Count frequencies
lang_counts = desire['LangList'].value_counts().reset_index()
lang_counts.columns = ['language','count']


# Save CSV
import os
os.makedirs('../output', exist_ok=True)
lang_counts.to_csv('../output/q4_desired_lang_2020.csv', index=False)

# Print top 10 and the top-most answer
print("Top 10 desired languages for 2020:")
print(lang_counts.head(10))


most_desired = lang_counts.iloc[0]['language'] if not lang_counts.empty else None
print("\nMost desired language for 2020 (survey):", most_desired)
```

```python
# Q5
import pandas as pd
import numpy as np
import country_converter as coco

df = pd.read_csv('../data/survey_results_public.csv', low_memory=False)
df = df.dropna()

# --- Normalize Gender ---
def normalize_gender(val):
    if pd.isna(val):
        return 'OTHERS'
    s = str(val).strip()
    if s == 'Man':
        return 'MAN'
    elif s == 'Woman':
        return 'WOMAN'
    else:
        return 'OTHERS'

# Apply
df['GenderClean'] = df['Gender'].apply(normalize_gender)

# --- Country -> Continent ---
cc = coco.CountryConverter()
df = df[df['Country'] != 'Other Country (Not Listed Above)']
df['Continent'] = cc.convert(names=df['Country'], to='continent', not_found='Unknown')

# --- Hobbyist ---
df['HobbyistFlag'] = df['Hobbyist'].map(lambda x: True if str(x).strip() == 'Yes' else False)

# --- Grouping ---
hobby_report = df.groupby(['Continent','GenderClean']).agg(
    total_respondents=('Respondent','nunique'),
    hobbyists=('HobbyistFlag','sum')
).reset_index()

hobby_report['pct_hobbyist'] = (hobby_report['hobbyists'] / hobby_report['total_respondents'] * 100).round(2)

print(hobby_report.head(10))
```

```python
# Q6
# --- Normalize Gender ---
def normalize_gender(val):
    if pd.isna(val):
        return 'OTHERS'
    s = str(val).strip()
    if s == 'Man':
        return 'MAN'
    elif s == 'Woman':
        return 'WOMAN'
    else:
        return 'OTHERS'
    
import pandas as pd
import numpy as np
import country_converter as coco

df = pd.read_csv('../data/survey_results_public.csv', low_memory=False)

# Create a working dataframe
job_carr_df = df[['Gender', 'Country', 'JobSat', 'CareerSat']].copy()
job_carr_df = job_carr_df.dropna(subset=['Country', 'JobSat', 'CareerSat'])

# Apply
job_carr_df['Gender'] = job_carr_df['Gender'].apply(normalize_gender)

# --- Country -> Continent ---
cc = coco.CountryConverter()
job_carr_df = job_carr_df[job_carr_df['Country'] != 'Other Country (Not Listed Above)']
job_carr_df['Continent'] = cc.convert(names=job_carr_df['Country'], to='continent', not_found='Unknown')

# Mapping satisfaction to numeric
satisfaction_map = {
    'Very satisfied': 5,
    'Slightly satisfied': 4,
    'Neither satisfied nor dissatisfied': 3,
    'Slightly dissatisfied': 2,
    'Very dissatisfied': 1
}
job_carr_df['JobSat'] = job_carr_df['JobSat'].map(satisfaction_map)
job_carr_df['CareerSat'] = job_carr_df['CareerSat'].map(satisfaction_map)

# Group by Gender & Continent, calculate mean satisfaction
job_carr_report = job_carr_df.groupby(['Continent', 'Gender']).agg({
    'JobSat': 'mean',
    'CareerSat': 'mean',
}).round(2)

job_carr_report
```
