```python
# Generate the report for the average salary of developers based on the continent.

import pandas as pd
import country_converter as coco

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')

# Keep only salary and country, drop missing values
df = df[['Country', 'ConvertedComp']].dropna()

# Exclude 'Other Country (Not Listed Above)'
df = df[df['Country'] != 'Other Country (Not Listed Above)']

# Map countries to continents
cc = coco.CountryConverter()
df['Continent'] = cc.convert(names=df['Country'], to='continent')

# Group by continent and calculate average salary
result = df.groupby('Continent').agg(
    Average_salary=('ConvertedComp', 'mean')
).reset_index()

# Sort by average salary descending
result_sorted = result.sort_values('Average_salary', ascending=False).reset_index(drop=True)

# Save CSV
import os
os.makedirs('../output', exist_ok=True)
result_sorted.to_csv('../output/q3_avg_salary_per_continent.csv', index=False)

# Display the top 10 continents by average salary
result_sorted.head(10)
```
