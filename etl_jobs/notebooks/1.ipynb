```python
# Find the average age of developers when they wrote their first line of code.
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('../data/survey_results_public.csv')

# Clean Age1stCode column
def clean_age(value):
    if pd.isna(value):
        return np.nan
    # Younger than 5 years
    if "Younger" in str(value):
        return 4
    # Older than 85
    if "Older" in str(value):
        return 86
    # Non-numeric values
    try:
        return int(value)
    except:
        return np.nan

# Apply cleaning function and drop NaN
df['Age1stCode_clean'] = df['Age1stCode'].apply(clean_age)
df = df.dropna(subset=['Age1stCode_clean'])

# Calculate average age
average_age = df['Age1stCode_clean'].mean()
print("Average age when developers wrote first line of code:", round(average_age, 2))
```
